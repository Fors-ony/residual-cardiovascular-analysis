```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# kaggle kernels pull aamir28/cardiovascular-risk-factor-analysis
```

```python
df= pd.read_csv("heart_disease_risk_2026.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing Values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
lipids=["cholesterol_total", "hdl", "ldl", "triglycerides"]
df[lipids].describe()
```

```python
df["tg_mmol"] = df["triglycerides"] / 88.57
df["hdl_mmol"] = df["hdl"] / 38.67
df[["triglycerides", "tg_mmol", "hdl", "hdl_mmol"]].head()
```

```python
df["AIP"]= np.log10(df["tg_mmol"] / df["hdl_mmol"])
df[["triglycerides", "hdl", "AIP"]].head()
```

```python
df["AIP"].describe()
```

```python
df["AIP_risk"]= pd.cut(
    df["AIP"],
    bins=[-np.inf, 0.11, 0.21, np.inf],
    labels=["Low Risk", "Intermediate Risk", "High Risk"]
)
df["AIP_risk"].value_counts()
```

```python
df["AIP"].describe()
```

```python
counts=df["AIP_risk"].value_counts().sort_index()
percent=(df["AIP_risk"].value_counts(normalize=True).sort_index() * 100).round(1)
pd.DataFrame({"count": counts, "percent": percent})
```

```python
lipid_cols=["ldl", "hdl", "triglycerides", "cholesterol_total"]
df.groupby("AIP_risk", observed=True)[lipid_cols].mean().round(1)
```

```python
(df.groupby("AIP_risk", observed=True)["has_heart_disease"].mean()* 100).round(1)
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df["AIP"], bins=40, color="steelblue", edgecolor="white")
plt.axvline(0.11, color="orange", linestyle="--", label="0.11 cutoff")
plt.axvline(0.21, color="red", linestyle="--", label="0.21 cutoff")
plt.xlabel("Atherogenic Index of Plasma (AIP)")
plt.ylabel("Number of patients")
plt.title("Distribution of AIP (synthetic dataset, n = 9,000)")
plt.legend()
plt.tight_layout()
plt.savefig("aip_distribution.png", dpi=300)
plt.show()
```

The histogram is roughly bell shaped and cemtered near 0.08 with the two cutoff lines marking the risk groups

```python
order = ["Low Risk", "Intermediate Risk", "High Risk"]
rates = df.groupby("AIP_risk", observed=True)["has_heart_disease"].mean() * 100
rates = rates.reindex(order)

plt.figure(figsize=(7, 4))
bars = plt.bar(order, rates.values, color=["#4caf50", "#ffb300", "#e53935"])
plt.bar_label(bars, fmt="%.1f%%")
plt.xlabel("AIP risk group")
plt.ylabel("Heart disease prevalence (%)")
plt.title("Heart disease prevalence by AIP risk group")
plt.tight_layout()
plt.savefig("heart_disease_by_aip.png", dpi=300)
plt.show()
```

The bar chart shows heart disease prevelance rising from 23.3% to 32.3% to 46.2%

```python
df["remnant_c"]= df["cholesterol_total"]-df["ldl"]-df["hdl"]
df["non_hdl"]= df["cholesterol_total"]-df["hdl"]
df["remnant_c"].describe()
```

```python
low_ldl = df[df["ldl"] < 100].copy()
print("People with LDL < 100 mg/dL:", len(low_ldl))

low_ldl["high_tg"] = low_ldl["triglycerides"] >= 150
low_ldl["high_remnant"] = low_ldl["remnant_c"] >= 30
low_ldl["high_nonhdl"] = low_ldl["non_hdl"] >= 130
flags = ["high_tg", "high_remnant", "high_nonhdl"]
low_ldl["any_abnormal"] = low_ldl[flags].any(axis=1)

pct = (low_ldl[flags + ["any_abnormal"]].mean() * 100).round(1)
pct
```

```python
labels = ["TG ≥150", "Remnant-C ≥30", "Non-HDL-C ≥130", "Any of the three"]
plt.figure(figsize=(8, 4.5))
bars = plt.bar(labels, pct.values, color=["#5c6bc0", "#26a69a", "#ffa726", "#e53935"])
plt.bar_label(bars, fmt="%.1f%%")
plt.ylabel("% of people with LDL-C < 100 mg/dL")
plt.title("Residual lipid abnormalities despite low LDL-C (synthetic data)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("residual_risk_low_ldl.png", dpi=300)
plt.show()
```
